### The setting

$$p(x;a,b) = H(a,b) \begin{cases} \frac{x}{a} & \text{if } 0 \leq x \leq a, \\ \frac{b-x}{b-a} & \text{if } a \leq x \leq b, \\ 0 & \text{otherwise.} \end{cases}$$
where $H(a,b)$ is a normalization constant. Alex has choosen $a$ and $b$ such that they are the first two and the last two digits of his phone secret code, respectively.

## Part 1: Properties of the distribution

The distribution $p(x;a,b)$ is a valid distribution, i.e., it is normalized and it is non-negative. In this part, you will find many properties of the distribution.

 - **Question 1.** Find the normalization constant $H(a,b)$.

We require 
$$
\int_{-\infty}^{\infty} p(x;a,b)\,dx = 1.
$$  
From this condition we will be able to find normalization constant. This further reduces to  
$$
\int_0^a \tfrac{x}{a}\,dx + \int_a^b \tfrac{b-x}{b-a}\,dx = \tfrac{a}{2} + \tfrac{b-a}{2} = \tfrac{b}{2}.
$$  
Thus the normalization constant is  
$$
H(a,b) = \frac{1}{b/2} = \frac{2}{b}.
$$

**Note.**  
We will assume both $a, b \neq 0$ and $a \neq b$, since otherwise the probability density function wouldn't be meaningful.

- **Question 2.** Compute the expected value $\mu$ of the distribution.

Be the definition of the expected value we can write 
$$
\mu = \int_{-\infty}^{\infty} x\,p(x;a,b)\,dx 
= \frac{2}{b}\left[ \int_0^a x\cdot\frac{x}{a}\,dx + \int_a^b x\cdot\frac{b-x}{b-a}\,dx \right].
$$  

Evaluating,  
$$
\int_0^a \frac{x^2}{a}\,dx = \frac{a^2}{3}, 
\quad
\int_a^b \frac{x(b-x)}{b-a}\,dx = \frac{b(b+a)}{6}.
$$  

Finally, we obtain the solution 
$$
\mu = \frac{2}{b}\left( \frac{a^2}{3} + \frac{b(b+a)}{6} - \frac{a^2}{2} \right) = \frac{a+b}{3}.
$$


- **Question 3.** Compute the _cumulative density function_ $F(t)$ of the distribution.

By the definition of the cumulative distribution function $\big(F(t) = P(X \leq t))$  we can write
$$
F(t)=\int_{-\infty}^t p(x;a,b)\,dx = 
\begin{cases}
0, & t<0, \\[6pt]
\frac{2}{b}\int_0^t \tfrac{x}{a}\,dx = \tfrac{t^2}{ab}, & 0\leq t \leq a, \\[8pt]
\frac{2}{b}\!\left(\int_0^a \tfrac{x}{a}\,dx + \int_a^t \tfrac{b-x}{b-a}\,dx\right)
= \tfrac{2bt - t^2 - ab}{b(b-a)}, & a \leq t \leq b, \\[8pt]
1, & t \geq b.
\end{cases}
$$

More detailed derviation for $a \leq t \leq b$,  
$$
F(t) = \frac{2}{b}\left[\int_0^a \frac{x}{a}\,dx + \int_a^t \frac{b-x}{b-a}\,dx\right].
$$  

The first integral is $\tfrac{a}{2}$, and the second is, with simple algebra steps 
$$
\frac{1}{b-a}\int_a^t (b-x)\,dx 
= \frac{bt - \tfrac{t^2}{2} - (ba - \tfrac{a^2}{2})}{b-a}.
$$  

Thus,  
$$
F(t) = \frac{2}{b}\cdot \frac{bt - \tfrac{t^2}{2} - \tfrac{ab}{2}}{b-a}
= \frac{2bt - t^2 - ab}{b(b-a)}.
$$


* **Question 4.** Compute the median $m$ of the distribution. (Be careful, there are some special cases to consider according to the value $a$ and $b$)

### Finding the Median from the CDF

We know that the median $m$ satisfies the condition (by definition of the median)  
$$
F(m) = \tfrac{1}{2}.
$$

We’ll find $m$ by solving this equation in each possible region of the CDF.


### 1) **$0 \le m \le a$**

In this range, the CDF is given by  
$$
F(m) = \frac{m^2}{ab}.
$$

Setting $F(m) = \tfrac{1}{2}$ gives  
$$
\frac{m^2}{ab} = \tfrac{1}{2}.
$$

Multiplying both sides by $ab$ and taking the positive square root (since $m \ge 0$) leads us to:
$$
m^2 = \tfrac{1}{2}ab
$$ 
$$
m = \sqrt{\tfrac{ab}{2}}.
$$

But this expression is only valid if $m$ actually lies in the interval $[0,a]$.  
That means we must have  
$$
\sqrt{\tfrac{ab}{2}} \le a.
$$

Squaring both sides:  
$$
\tfrac{ab}{2} \le a^2 \quad \Rightarrow \quad b \le 2a.
$$

So this formula for $m$ is valid whnever $b \le 2a$.


### 2) **$a \le m \le b$**

For this region, the CDF is  
$$
F(m) = \frac{2bm - m^2 - ab}{b(b-a)}.
$$

Setting $F(m) = \tfrac{1}{2}$ gives  
$$
\frac{2bm - m^2 - ab}{b(b-a)} = \tfrac{1}{2}.
$$

Multiplying both sides by $b(b-a)$ to remove the denominator:  
$$
2bm - m^2 - ab = \tfrac{1}{2}b(b-a)
$$ 
$$
m^2 - 2bm + \tfrac{b^2 + ab}{2} = 0.
$$

Now solving the quadratic formula for $m$:  
$$
m = \frac{2b \pm \sqrt{(2b)^2 - 4 \cdot 1 \cdot \tfrac{b^2 + ab}{2}}}{2}
$$

$$
m = b \pm \sqrt{b^2 - \tfrac{b^2 + ab}{2}}
$$

$$
\big(b^2 - \tfrac{b^2 + ab}{2} = \tfrac{2b^2 - b^2 - ab}{2} = \tfrac{b(b - a)}{2})
$$
 
$$
m = b \pm \sqrt{\tfrac{b(b - a)}{2}}.
$$

Since the median must lie in $[a, b]$, we take the smaller root (which is for sure less than $b$):  
$$
m = b - \sqrt{\tfrac{b(b - a)}{2}}.
$$

Now we must check the validity of this expression — it should lie in the region $[a, b]$ (upper bound is ensured by taking the solution of the quadratic equation which is lesser than $b$).  
Now, we also require  
$$
m \ge a.
$$

Substitute $m = b - \sqrt{\tfrac{b(b - a)}{2}}$ into this inequality:
$$
b - \sqrt{\tfrac{b(b - a)}{2}} \ge a
$$
We can square both sides (since both are positive for $b > a$) and divide by $(b - a)$ (valid for $b > a$):
$$
b - a \ge \sqrt{\tfrac{b(b - a)}{2}}
$$


$$
(b - a)^2 \ge \tfrac{b(b - a)}{2}
$$
$$
b - a \ge \tfrac{b}{2}
$$
$$
b - \tfrac{b}{2} \ge a \quad \Rightarrow \quad \tfrac{b}{2} \ge a
$$

So this condition is equivalent to  
$$
b \ge 2a.
$$

Thus, this case is valid when $b > 2a$ (strict inequality since we switched regions from 1)).

### End expression
Finally, we combine everything: 

$$
m =
\begin{cases}
\sqrt{\tfrac{ab}{2}}, & b \le 2a, \\[10pt]
b - \sqrt{\tfrac{b(b - a)}{2}}, & b > 2a.
\end{cases}
$$

- **Question 6.** Implement a function `pdf(x, a, b=20)` that computes the probability density function $p(x;a,20)$, and plot it for $b=20$ and $a$ of your choice.

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.optimize import minimize

def pdf(x, a, b=20):
    """
    Piecewise probability density function (PDF).

    Parameters
    ----------
    x : array_like
        Input values where the PDF is evaluated.
    a : float
        Shape parameter. Must satisfy 0 < a < b.
    b : float, optional
        Upper bound of the distribution (default = 20). Must be > a.

    Returns
    -------
    pdfs : ndarray
        Array of the same shape as `x`, containing the PDF values.
    """
    # Ensure numpy array
    x = np.asarray(x, dtype=float)

    # Return zeros if parameters are invalid
    if a <= 0 or a >= b:
        return 1e-300*np.zeros_like(x, dtype=float)

    interval1 = (x >= 0) & (x < a)
    interval2 = (x >= a) & (x < b)

    pdfs = np.zeros_like(x, dtype=float)
    pdfs[interval1] = (x[interval1] / a)
    pdfs[interval2] = ((b - x[interval2]) / (b - a))
    pdfs *= (2.0 / b)

    return pdfs

# plotting
a = 15
x_vals = np.linspace(-5, 25, 200)
y_vals = pdf(x_vals, a)
plt.figure()
plt.plot(x_vals, y_vals)
plt.xlabel('x'); plt.ylabel('p(x)')
plt.title('PDF'); plt.grid(True)
plt.show()
```

 - **Question 14.** Compute the Fisher information of $a$ under the distribution $p(x;a,b=20)$.


We start the definition of Fisher Information
$$
I(a) \;=\; \mathbb{E}\!\left[\left(\frac{\partial}{\partial a}\log p(X;a,b)\right)^2\right].
$$

For $0 \leq x \leq a$:
$$
\log p(x;a,b) = \log\!\left(\tfrac{2}{b}\cdot \tfrac{x}{a}\right), 
\qquad \frac{\partial}{\partial a}\log p(x;a,b) = -\tfrac{1}{a}.
$$

For $a \leq x \leq b$:
$$
\log p(x;a,b) = \log\!\left(\tfrac{2}{b}\cdot \tfrac{b-x}{b-a}\right), 
\qquad \frac{\partial}{\partial a}\log p(x;a,b) = \tfrac{1}{b-a}.
$$

Thus, the squared derivative is piecewise constant:
$$
\left(\tfrac{\partial}{\partial a}\log p(x;a,b)\right)^2 =
\begin{cases}
\tfrac{1}{a^2}, & 0 \leq x \leq a, \\[6pt]
\tfrac{1}{(b-a)^2}, & a \leq x \leq b.
\end{cases}
$$

$$
I(a) \;=\; \int_0^a \frac{1}{a^2}\,p(x;a,b)\,dx \;+\; \int_a^{b} \frac{1}{(b-a)^2}\,p(x;a,b)\,dx.
$$

- First part:
$$
\int_0^a \frac{1}{a^2}\,\frac{2}{b}\frac{x}{a}\,dx
= \frac{2}{ba^3}\int_0^a x\,dx
= \frac{1}{ba}.
$$

- Second part:
$$
\int_a^{b} \frac{1}{(b-a)^2}\,\frac{2}{b}\frac{b-x}{b-a}\,dx
= \frac{2}{b(b-a)^3}\int_a^{b}(b-x)\,dx
= \frac{1}{b(b-a)}.
$$

**Final expression**

Combining both parts gives:
$$
I(a,b) = \frac{1}{b}\!\left(\frac{1}{a} + \frac{1}{b-a}\right)
       = \frac{1}{a(b-a)}, \qquad 0<a<b.
$$

For $b=20$, this becomes
$$
I(a) = \frac{1}{a(20-a)}.
$$

And for the $n$ samples, as they are IID, we simply sum this expression $n$ times. 

$$
I(a) = \frac{n}{a(20-a)}.
$$

### Note on regularity conditions
The alternative formula
$$
I(a) = -\mathbb{E}\!\left[\frac{\partial^2}{\partial a^2}\log p(X;a,b)\right]
$$
is **not valid here**, because the support of $p(x;a,b)$ depends on $a$ (the boundary at $x=a$ moves with the parameter).  
This violates the regularity conditions, and additional boundary terms appear when integrating by parts.  
Therefore, only the first definition (squared derivative of the log-likelihood) should be used.

```python
def prior_jeff(a):
    x = (1/20) * (1/(a + 1e-10) + 1/(20 - a + 1e-10))
    return np.sqrt(np.clip(x, 0, None))

def loglh_jeff(x, a, b=20):
    return loglh(x, a, b) + np.log(len(x)*prior_jeff(a))
```

```python
def est_a_jeff(data):
    a_guess = 10
    result = minimize(lambda a: -loglh_jeff(data, a), a_guess, method='Nelder-Mead')
    return result.x[0]
```

### Test the estimators
You are still trying to sneak into your friend's computer to get new samples of the distribution. In the meanwhile, you want to test the estimators you derived on a synthetic dataset. Florent gave you this Python function to sample the distribution $p(x;a,b)$:


```python
import numpy as np
import matplotlib.pyplot as plt

def sample(a: float, b: float, size: int=1) -> np.array:
    def sample_single(a: float, b: float) -> float:
        u = 0
        v = b
        while not((u <= a and v <= 2/(a*b)*u) or (u > a and v <= 2/b *(b-u)/(b-a))):
            u = np.random.uniform(0, b)
            v = np.random.uniform(0, 2/b)
        return u
    return np.array([sample_single(a, b) for _ in range(size)])
```

- **Bonus question 1** The sampling function is not efficient. Can you provide a more efficient way to sample the distribution?

We want to sample from the triangular distribution on $[0,b]$ with mode $a$.  
The idea is the inverse CDF method:  
if we know the CDF $F(x)$, then for $U \sim \text{Uniform}(0,1)$ we can set  
$$X = F^{-1}(U),$$  
and $X$ will follow the same distribution as $F$

Let's start from the CDF that we derived previously:
$$
F(x) =
\begin{cases}
\dfrac{x^2}{a b}, & 0 \le x \le a, \\[6pt]
1 - \dfrac{(b-x)^2}{b(b-a)}, & a < x \le b.
\end{cases}
$$

For the first part, $0 \le x \le a$:
$$u = \frac{x^2}{a b} \quad \Longrightarrow \quad x = \sqrt{u a b}.$$

For the second part, $a < x \le b$:
$$u = 1 - \frac{(b-x)^2}{b(b-a)} \quad \Longrightarrow \quad x = b - \sqrt{(1-u)b(b-a)}.$$

At the point $x=a$, the corresponding $u$ value is $F(a)=a/b$, which is where the two formulas meet.  
After combining them we get the full inverse function:
$$
X =
\begin{cases}
\sqrt{U a b}, & U < a/b, \\[8pt]
b - \sqrt{(1 - U)\,b(b - a)}, & U \ge a/b.
\end{cases}
$$

So, to sum up, we sample from a $U \sim \text{Uniform}(0,1)$ distribution and see in which  range obtained sample $U$ belongs, and depending on that we calculate the "inverse CDF" using an appropriate formula.

```python
def optimized_sampling(a: float, b: float, size: int=1) -> np.array:
    if a <= 0 or b <= a:
        raise RuntimeError("Invalid input!")
    u = np.random.rand(size)
    condition = u < a/b
    x = condition * np.sqrt(u*a*b) + (~condition)*(b - np.sqrt((1-u)*b*(b-a)))
    return x
```

We will first recall the mean and the median which we calculated in the very beginning
$$
\mu=\frac{a+b}{3},
\qquad
m=

\sqrt{\dfrac{ab}{2}}, & b\le 2a,\\[6pt]

My idea was to do the similar thing to the case when $b$ was known, and we tried to find the estimator for $a$ using mean and median. Here we would just solve the system of equations (over $a$ and $b$) taking into account both the mean and median. Thus, we will solve a system of equations.

### 1) $b\le 2a$

Here $m=\sqrt{ab/2}\Rightarrow ab=2m^2$, and $\mu=(a+b)/3\Rightarrow b=3\mu-a$.

Substitute $b=3\mu-a$ into $ab=2m^2$:
$$
a(3\mu-a)=2m^2
\;\Longrightarrow\;
-a^2+3\mu a-2m^2=0.
$$
This is a quadratic in $a$. Solve:
$$
a=\frac{3\mu\pm\sqrt{9\mu^2-8m^2}}{2}.
$$
Because $a\le b$ and $a+b=3\mu$, the smaller root is $a$ and the larger is $b$ (after writing $b$ in terms of $a$ using $a+b=3\mu$ we observe that the two possible values for $b$ are the same as for $a$):
$$
\boxed{
a=\frac{3\mu-\sqrt{9\mu^2-8m^2}}{2},\qquad
b=\frac{3\mu+\sqrt{9\mu^2-8m^2}}{2}
}.
$$
Feasibility requires the discriminant $9\mu^2-8m^2\ge 0$.
After computing $(a,b)$, check the case condition $b\le 2a$; if it fails, switch to 2).

### 2) $b>2a$

Here $m=b-\sqrt{b(b-a)/2}$. Let $s=b-m$ (the positive square-root term). Then
$$
s=\sqrt{\frac{b(b-a)}{2}}
\;\Longrightarrow\;
(b-m)^2=\frac{b(b-a)}{2}.
$$
Expand and clear the denominator:
$$
b^2-2bm+m^2=\frac{b^2}{2}-\frac{ab}{2}
\;\Longrightarrow\;
b^2-4bm+2m^2+ab=0.
$$
Use $\mu=(a+b)/3\Rightarrow a=3\mu-b$ and substitute:
$$
b^2-4bm+2m^2+b(3\mu-b)=0
\;\Longrightarrow\;
(4m-3\mu)b-2m^2=0.
$$
Solve for $b$:
$$
\boxed{b=\frac{2m^2}{\,4m-3\mu\,}}.
$$
Then recover $a$ from $\mu$:
$$
\boxed{a=3\mu-b}.
$$
Feasibility requires $4m-3\mu>0$ (so the denominator is positive) and, after computing, verify the case condition $b>2a$.



Now, with these expressions, we have the estimators and should just put the hats on relevant symbols. So, in general case we should compute, for example, first 1) (but previously check for feasibility with relation to mean and median) and check if the condition $b \leq 2a$. If not, we compute second case (again, if the feasibility holds). 

Here spot a couple of drawbacks. When we have only a couple of samples, there could be no feasible solution, no real $a$ and $b$ that produced such $median$ and $mean$. Secondly, for discontinuos case switch (near $b=2a$) we can get unstable results.

On the other hand, the estimators are consistent (by the Law of Large Numbers), with infinite numbers of samples, mean and median will converge towards their true values, and because they are mapped into our parameters continually, we will get true uknown parameters as well.

So, to conclude, I would expect these estimators to work with a lot of samples.
