$$p(x;a,b) = H(a,b) \begin{cases} \frac{x}{a} & \text{if } 0 \leq x \leq a, \\ \frac{b-x}{b-a} & \text{if } a \leq x \leq b, \\ 0 & \text{otherwise.} \end{cases}$$

## Part 1: Properties of the distribution

The distribution $p(x;a,b)$ is a valid distribution, i.e., it is normalized and it is non-negative. In this part, you will find many properties of the distribution.

 - **Question 1.** Find the normalization constant $H(a,b)$.

We require  
$$
\int_{-\infty}^{\infty} p(x;a,b)\,dx = 1.
$$  
This reduces to  
$$
\int_0^a \tfrac{x}{a}\,dx + \int_a^b \tfrac{b-x}{b-a}\,dx = \tfrac{a}{2} + \tfrac{b-a}{2} = \tfrac{b}{2}.
$$  
Hence the normalization constant is  
$$
H(a,b) = \frac{1}{b/2} = \frac{2}{b}.
$$

**Note.**  
We will assume both $a, b \neq 0$ and $a \neq b$, since otherwise the probability density function wouldn't be meaningful.

- **Question 2.** Compute the expected value $\mu$ of the distribution.

We compute  
$$
\mu = \int_{-\infty}^{\infty} x\,p(x;a,b)\,dx 
= \frac{2}{b}\left[ \int_0^a x\cdot\frac{x}{a}\,dx + \int_a^b x\cdot\frac{b-x}{b-a}\,dx \right].
$$  

Evaluating,  
$$
\int_0^a \frac{x^2}{a}\,dx = \frac{a^2}{3}, 
\quad
\int_a^b \frac{x(b-x)}{b-a}\,dx = \frac{b(b+a)}{6}.
$$  

Thus,  
$$
\mu = \frac{2}{b}\left( \frac{a^2}{3} + \frac{b(b+a)}{6} - \frac{a^2}{2} \right) = \frac{a+b}{3}.
$$


- **Question 3.** Compute the _cumulative density function_ $F(t)$ of the distribution.

We compute  
$$
F(t)=\int_{-\infty}^t p(x;a,b)\,dx = 
\begin{cases}
0, & t<0, \\[6pt]
\frac{2}{b}\int_0^t \tfrac{x}{a}\,dx = \tfrac{t^2}{ab}, & 0\leq t \leq a, \\[8pt]
\frac{2}{b}\!\left(\int_0^a \tfrac{x}{a}\,dx + \int_a^t \tfrac{b-x}{b-a}\,dx\right)
= \tfrac{2bt - t^2 - ab}{b(b-a)}, & a \leq t \leq b, \\[8pt]
1, & t \geq b.
\end{cases}
$$

For $a \leq t \leq b$,  
$$
F(t) = \frac{2}{b}\left[\int_0^a \frac{x}{a}\,dx + \int_a^t \frac{b-x}{b-a}\,dx\right].
$$  

The first integral is $\tfrac{a}{2}$, and the second is  
$$
\frac{1}{b-a}\int_a^t (b-x)\,dx 
= \frac{bt - \tfrac{t^2}{2} - (ba - \tfrac{a^2}{2})}{b-a}.
$$  

Thus,  
$$
F(t) = \frac{2}{b}\cdot \frac{bt - \tfrac{t^2}{2} - \tfrac{ab}{2}}{b-a}
= \frac{2bt - t^2 - ab}{b(b-a)}.
$$


* **Question 4.** Compute the median $m$ of the distribution. (Be careful, there are some special cases to consider according to the value $a$ and $b$)

The median $m$ is defined by  
$$
F(m) = \tfrac{1}{2}.
$$  

**Case 1: $0 \leq m \leq a$**  
From the CDF,  
$$
F(m) = \frac{m^2}{ab}.
$$  
So the condition $F(m) = \tfrac{1}{2}$ gives  
$$
m = \sqrt{\tfrac{ab}{2}}, \quad \text{valid if } m \leq a \;\;\Leftrightarrow\;\; b \leq 2a.
$$  

**Case 2: $a \leq m \leq b$**  
From the CDF,  
$$
F(m) = \frac{2bm - m^2 - ab}{b(b-a)}.
$$  
Setting $F(m) = \tfrac{1}{2}$ leads to  
$$
2bm - m^2 - ab = \tfrac{1}{2} b(b-a).
$$  
Rearranging gives  
$$
m^2 - 2bm + \tfrac{b^2+ab}{2} = 0,
$$  
with solution inside $[a,b]$  
$$
m = b - \sqrt{\tfrac{b(b-a)}{2}}, \quad \text{valid if } b > 2a.
$$  

**Final Answer**  
$$
m =
\begin{cases}
\sqrt{\tfrac{ab}{2}}, & b \leq 2a, \\[10pt]
b - \sqrt{\tfrac{b(b-a)}{2}}, & b > 2a.
\end{cases}
$$

- **Question 6.** Implement a function `pdf(x, a, b=20)` that computes the probability density function $p(x;a,20)$, and plot it for $b=20$ and $a$ of your choice.

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.optimize import minimize

def pdf(x, a, b=20):
    """
    Piecewise probability density function (PDF).

    Parameters
    ----------
    x : array_like
        Input values where the PDF is evaluated.
    a : float
        Shape parameter. Must satisfy 0 < a < b.
    b : float, optional
        Upper bound of the distribution (default = 20). Must be > a.

    Returns
    -------
    pdfs : ndarray
        Array of the same shape as `x`, containing the PDF values.
    """
    # Ensure numpy array
    x = np.asarray(x, dtype=float)

    # Return zeros if parameters are invalid
    if a <= 0 or a >= b:
        return 1e-300*np.zeros_like(x, dtype=float)

    interval1 = (x >= 0) & (x < a)
    interval2 = (x >= a) & (x < b)

    pdfs = np.zeros_like(x, dtype=float)
    pdfs[interval1] = (x[interval1] / a)
    pdfs[interval2] = ((b - x[interval2]) / (b - a))
    pdfs *= (2.0 / b)

    return pdfs

# plotting
a = 15
x_vals = np.linspace(-5, 25, 200)
y_vals = pdf(x_vals, a)
plt.figure()
plt.plot(x_vals, y_vals)
plt.xlabel('x'); plt.ylabel('p(x)')
plt.title('PDF'); plt.grid(True)
plt.show()
```

 - **Question 14.** Compute the Fisher information of $a$ under the distribution $p(x;a,b=20)$.

### Fisher Information of $a$ (with $b$ constant)

We use the definition
$$
I(a) \;=\; \mathbb{E}\!\left[\left(\frac{\partial}{\partial a}\log p(X;a,b)\right)^2\right].
$$

---

**Derivative of the log-density**

For $0 \leq x \leq a$:
$$
\log p(x;a,b) = \log\!\left(\tfrac{2}{b}\cdot \tfrac{x}{a}\right), 
\qquad \frac{\partial}{\partial a}\log p(x;a,b) = -\tfrac{1}{a}.
$$

For $a \leq x \leq b$:
$$
\log p(x;a,b) = \log\!\left(\tfrac{2}{b}\cdot \tfrac{b-x}{b-a}\right), 
\qquad \frac{\partial}{\partial a}\log p(x;a,b) = \tfrac{1}{b-a}.
$$

Thus, the squared derivative is piecewise constant:
$$
\left(\tfrac{\partial}{\partial a}\log p(x;a,b)\right)^2 =
\begin{cases}
\tfrac{1}{a^2}, & 0 \leq x \leq a, \\[6pt]
\tfrac{1}{(b-a)^2}, & a \leq x \leq b.
\end{cases}
$$

---

**Expectation over each region**

$$
I(a) \;=\; \int_0^a \frac{1}{a^2}\,p(x;a,b)\,dx \;+\; \int_a^{b} \frac{1}{(b-a)^2}\,p(x;a,b)\,dx.
$$

- First part:
$$
\int_0^a \frac{1}{a^2}\,\frac{2}{b}\frac{x}{a}\,dx
= \frac{2}{ba^3}\int_0^a x\,dx
= \frac{1}{ba}.
$$

- Second part:
$$
\int_a^{b} \frac{1}{(b-a)^2}\,\frac{2}{b}\frac{b-x}{b-a}\,dx
= \frac{2}{b(b-a)^3}\int_a^{b}(b-x)\,dx
= \frac{1}{b(b-a)}.
$$

---

**Final expression**

Combining both parts gives:
$$
I(a,b) = \frac{1}{b}\!\left(\frac{1}{a} + \frac{1}{b-a}\right)
       = \frac{1}{a(b-a)}, \qquad 0<a<b.
$$

For $b=20$, this becomes
$$
I(a) = \frac{1}{a(20-a)}.
$$

---

### Note on regularity conditions
The alternative formula
$$
I(a) = -\mathbb{E}\!\left[\frac{\partial^2}{\partial a^2}\log p(X;a,b)\right]
$$
is **not valid here**, because the support of $p(x;a,b)$ depends on $a$ (the boundary at $x=a$ moves with the parameter).  
This violates the regularity conditions, and additional boundary terms appear when integrating by parts.  
Therefore, only the first definition (squared derivative of the log-likelihood) should be used.


- **Question 15.** Can you compute the maximum likelihood estimator using the Jeffrey Posterior $\hat a_\text{jeff}$? Provide also the confidence intervals.

```python
def prior_jeff(a):
    return np.sqrt(1/(20*a) + 1/(20*(20-a)))

def loglh_jeff(x, a, b=20):
    return loglh(x, a, b) + np.log(len(x)*prior_jeff(a))
```

You are still trying to sneak into your friend's computer to get new samples of the distribution. In the meanwhile, you want to test the estimators you derived on a synthetic dataset. Florent gave you this Python function to sample the distribution $p(x;a,b)$:


```python
import numpy as np
import matplotlib.pyplot as plt

def sample(a: float, b: float, size: int=1) -> np.array:
    def sample_single(a: float, b: float) -> float:
        u = 0
        v = b
        while not((u <= a and v <= 2/(a*b)*u) or (u > a and v <= 2/b *(b-u)/(b-a))):
            u = np.random.uniform(0, b)
            v = np.random.uniform(0, 2/b)
        return u
    return np.array([sample_single(a, b) for _ in range(size)])
```

- **Question 19.** Can you provide the confidence intervals for the estimators? Can you guess the secret code of your friend?

```python
np.random.seed(123)
n_boot = len(data)
n_repetitions = 100

a_bootstrap = []
b_bootstrap = []

for i in range(n_repetitions):
    data_boot = np.random.choice(data, n_boot, replace=True)
    a_ml, b_ml = est_ab_ml(data_boot)
    a_bootstrap.append(a_ml)
    b_bootstrap.append(b_ml)


a_bootstrap = np.array(a_bootstrap)
b_bootstrap = np.array(b_bootstrap)

# 95% CI
alpha = 0.05
ci_a = np.percentile(a_bootstrap, [100*alpha/2, 100*(1-alpha/2)])
ci_b = np.percentile(b_bootstrap, [100*alpha/2, 100*(1-alpha/2)])

print("95% CI for a_ML:", ci_a)
print("95% CI for b_ML:", ci_b)
```

```python
fig, axes = plt.subplots(1, 2, figsize=(13, 5))

# a_ML histogram
axes[0].hist(a_bootstrap, bins=15, color="cornflowerblue", edgecolor="black", alpha=0.75)
axes[0].axvline(ci_a[0], color="crimson", ls="--", lw=2, label="95% CI lower")
axes[0].axvline(ci_a[1], color="crimson", ls="--", lw=2, label="95% CI upper")
axes[0].set_xlabel(r"$\hat{a}_{ML}$", fontsize=13)
axes[0].set_ylabel("Frequency", fontsize=13)
axes[0].set_title("Bootstrap Distribution of $a_{ML}$", fontsize=14)
axes[0].legend(fontsize=10)
axes[0].grid(alpha=0.3)

# b_ML histogram
axes[1].hist(b_bootstrap, bins=15, color="mediumseagreen", edgecolor="black", alpha=0.75)
axes[1].axvline(ci_b[0], color="crimson", ls="--", lw=2, label="95% CI lower")
axes[1].axvline(ci_b[1], color="crimson", ls="--", lw=2, label="95% CI upper")
axes[1].set_xlabel(r"$\hat{b}_{ML}$", fontsize=13)
axes[1].set_ylabel("Frequency", fontsize=13)
axes[1].set_title("Bootstrap Distribution of $b_{ML}$", fontsize=14)
axes[1].legend(fontsize=10)
axes[1].grid(alpha=0.3)

plt.tight_layout()
plt.show()
```
